```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/DELL/Downloads/Sleep_health_and_lifestyle_dataset.csv")
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_sleep = df[df["Sleep Duration"] > 7]
print(high_sleep)
```

```python
female = df[df["Gender"] == "Female"]
print(female)
```

```python
sorted_df = df.sort_values(
    by="Sleep Duration",
    ascending=False
)
print(sorted_df)
```

```python
print("Mean =",
      df["Sleep Duration"].mean())
print("Median =",
      df["Sleep Duration"].median())
print("Mode =")
print(df["Sleep Duration"].mode())
print("Standard Deviation =",
      df["Sleep Duration"].std())
```

```python
print(df.describe())
```

```python
gender_avg = df.groupby(
    "Gender"
)["Sleep Duration"].mean()
print(gender_avg)
```

```python
corr = df[[
    "Sleep Duration",
    "Quality of Sleep"
]].corr()
print(corr)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.scatter(
    df["Sleep Duration"],
    df["Quality of Sleep"]
)
plt.xlabel("Sleep Duration")
plt.ylabel("Quality of Sleep")
plt.title("Sleep Duration vs Quality of Sleep")
plt.show()
```

```python
import matplotlib.pyplot as plt
top = df.sort_values(
    by="Sleep Duration",
    ascending=False
).head(10)
plt.figure(figsize=(8,5))
plt.bar(top["Person ID"].astype(str),
        top["Sleep Duration"])
plt.xlabel("Person ID")
plt.ylabel("Sleep Duration")
plt.title("Top 10 Sleep Duration")
plt.show()
```
