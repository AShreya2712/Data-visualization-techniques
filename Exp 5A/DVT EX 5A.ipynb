```python
import matplotlib.pyplot as plt
```

```python
months = ["January", "February", "March", "April", "May", "June"]
sales = [100, 150, 120, 180, 200, 250]

plt.plot(months, sales)

plt.title("Monthly Sales")

plt.xlabel("Months")

plt.ylabel("Sales")

plt.show()
```

```python
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()
```

```python
plt.hist(sales, bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.boxplot(sales)
plt.title("Sales Box Plot")
plt.ylabel("Sales")
plt.show()
```

```python
profit = [20, 30, 40, 35, 50, 60]
plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.title("Sales and Profit")
plt.xlabel("Months")
plt.ylabel("Value")
plt.legend()
plt.show()
```
