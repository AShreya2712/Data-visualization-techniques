```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,150,180,170,220,250]
```

```python
plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
