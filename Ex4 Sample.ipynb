```python
import pandas as pd
data={
    "Roll No":[101,102,103,104,105,106],
    "Name":["Arun","Bala","Divya","Priya","Ravi","Karthik"],
    "Department":["CSE","IT","CSE","AI&DS","IT","AI&DS"],
    "Marks":[85,78,95,90,72,88],
    "Attendance":[92,88,96,94,85,91]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_marks=df[df["Marks"]>85]
print(high_marks)
```

```python
it_students=df[df["Department"]=="IT"]
print(it_students)
```

```python
sorted_df = df.sort_values(
    by="Marks",
    ascending=False
)
print(sorted_df)
```

```python
print("Mean =",
      df["Marks"].mean())

print("Median =",
      df["Marks"].median())

print("Mode =",
      df["Marks"].mode())

print("Standard Deviation =",
      df["Marks"].std())
```

```python
print(df.describe())
```

```python
dept_avg = df.groupby(
    "Department"
)["Marks"].mean()

print(dept_avg)
```

```python
corr = df[[
"Marks",
"Attendance"
]].corr()

print(corr)
```
