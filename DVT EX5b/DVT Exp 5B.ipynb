```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("C:/Users/SHREYA/Downloads/Sleep_health_and_lifestyle_dataset.csv")
df.head()
```

```python
age_sleep = df.groupby("Age")["Sleep Duration"].mean()
plt.figure(figsize=(10,5))
plt.plot(age_sleep.index, age_sleep.values, marker='o')
plt.title("Average Sleep Duration by Age")
plt.xlabel("Age")
plt.ylabel("Average Sleep Duration (Hours)")
plt.grid(True)
plt.show()
```

```python
occupation_sleep = df.groupby("Occupation")["Sleep Duration"].mean()
plt.figure(figsize=(10,6))
plt.bar(occupation_sleep.index, occupation_sleep.values)
plt.title("Average Sleep Duration by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Sleep Duration (Hours)")
plt.xticks(rotation=45)
plt.show()
```

```python
sleep_disorder = df["Sleep Disorder"].fillna("None").value_counts()
plt.figure(figsize=(8,6))
plt.pie(
    sleep_disorder.values,
    labels=sleep_disorder.index,
    autopct='%1.1f%%'
)
plt.title("Sleep Disorder Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sleep Duration"], bins=10)
plt.title("Distribution of Sleep Duration")
plt.xlabel("Sleep Duration (Hours)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(
    df["Physical Activity Level"],
    df["Sleep Duration"],
    alpha=0.5
)
plt.title("Physical Activity Level vs Sleep Duration")
plt.xlabel("Physical Activity Level")
plt.ylabel("Sleep Duration (Hours)")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.boxplot(df["Sleep Duration"].dropna())
plt.title("Sleep Duration Box Plot")
plt.ylabel("Sleep Duration (Hours)")
plt.show()
```

```python
sleep_duration = df.groupby("Age")["Sleep Duration"].mean()

# Calculate average quality of sleep by age

sleep_quality = df.groupby("Age")["Quality of Sleep"].mean()

# Create multiple line chart

plt.figure(figsize=(10,5))

plt.plot(
    sleep_duration.index,
    sleep_duration.values,
    marker='o',
    label="Sleep Duration"
)

plt.plot(
    sleep_quality.index,
    sleep_quality.values,
    marker='s',
    label="Quality of Sleep"
)

plt.title("Sleep Duration and Quality of Sleep by Age")

plt.xlabel("Age")

plt.ylabel("Average Value")

plt.legend()

plt.grid(True)

plt.show()
```
